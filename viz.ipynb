```python
from datasets import load_dataset

dataset = load_dataset("uoft-cs/cifar10")

train_dataset = dataset["train"]
test_dataset = dataset["test"]

print(len(train_dataset))  # 50000
print(len(test_dataset))   # 10000
```

```python
import numpy as np

hf_train = dataset["train"]
hf_test = dataset["test"]

train_images = np.stack([
    np.array(img.convert("RGB"))
    for img in hf_train["img"]
]).astype(np.float32) / 255.0

train_labels = np.array(hf_train["label"])

test_images = np.stack([
    np.array(img.convert("RGB"))
    for img in hf_test["img"]
]).astype(np.float32) / 255.0

test_labels = np.array(hf_test["label"])
```

```python
import matplotlib.pyplot as plt
import numpy as np

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

fig, axes = plt.subplots(10, 3, figsize=(8, 20))

for class_idx in range(10):
    indices = np.where(train_labels == class_idx)[0]
    selected = np.random.choice(indices, 3, replace=False)

    for j, idx in enumerate(selected):
        axes[class_idx, j].imshow(train_images[idx])
        axes[class_idx, j].set_title(
            f"{class_names[train_labels[idx]]} (Label: {train_labels[idx]})"
        )
        axes[class_idx, j].axis("off")

plt.tight_layout()
plt.show()

plt.tight_layout()
plt.show()
```
